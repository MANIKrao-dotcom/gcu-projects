```python
accounts = {
    "Manik": 25000,
    "Taufiq": 12000,
    "Abhi": 45000,
    "Surya": 8000
}

total = 0
highest = 0
richest = ""

for name, balance in accounts.items():
    total += balance

    if balance > highest:
        highest = balance
        richest = name

print("Total Money :", total)
print("Richest Person :", richest)
print("Highest Balance :", highest)

print("\nLow Balance Accounts")
for name, balance in accounts.items():
    if balance < 10000:
        print(name)
```
