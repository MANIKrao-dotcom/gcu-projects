```python
students = [
    {"name": "Manik", "marks": 85, "age": 19},
    {"name": "Taufiq", "marks": 45, "age": 20},
    {"name": "Abhi", "marks": 72, "age": 18},
    {"name": "Surya", "marks": 90, "age": 21},
    {"name": "Deepu", "marks": 55, "age": 17}
]

result = filter(
    lambda x: x["age"] >= 18,
    filter(
        lambda x: x["marks"] >= 60,
        students
    )
)

print(list(result))
```
