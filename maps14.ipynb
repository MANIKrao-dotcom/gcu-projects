```python
employees = ["Manik", "Rahul", "Amit", "Taufiq", "Suresh"]

salary = [25000, 32000, 45000, 28000, 50000]

bonus = list(map(
    lambda x: x * 0.10,
    salary
))

hra = list(map(
    lambda x: x * 0.20,
    salary
))

tax = list(map(
    lambda x: x * 0.05 if x > 30000 else x * 0.02,
    salary
))

final_salary = list(map(
    lambda s, b, h, t: s + b + h - t,
    salary, bonus, hra, tax
))

print("EMPLOYEE SALARY REPORT")
print("==============================")

for i in range(len(employees)):
    print("Employee:", employees[i])
    print("Basic Salary:", salary[i])
    print("Bonus:", bonus[i])
    print("HRA:", hra[i])
    print("Tax:", tax[i])
    print("Final Salary:", final_salary[i])
    print("------------------------------")
```
