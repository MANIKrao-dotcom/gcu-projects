```python
products = (
    ("Laptop", (50000, 10)),
    ("Mobile", (25000, 15)),
    ("Tablet", (18000, 8))
)

for product in products:
    name = product[0]
    price = product[1][0]
    quantity = product[1][1]

    total = price * quantity

    print("Product:", name)
    print("Price:", price)
    print("Quantity:", quantity)
    print("Total Value:", total)
    print()
```
