```python
def calculate_average(numbers):
    def is_valid(num):
        return isinstance(num, (int, float))

    clean_nums = [n for n in numbers if is_valid(n)]
    return sum(clean_nums) / len(clean_nums) if clean_nums else 0

print(calculate_average([10, "invalid", 20, 30]))  # Output: 20.0
```
