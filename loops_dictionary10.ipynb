```python
cart = {
    "Laptop": 55000,
    "Mouse": 800,
    "Keyboard": 1500,
    "Headphones": 2000
}

total = 0
costliest = ""
high = 0

for item, price in cart.items():
    total += price

    if price > high:
        high = price
        costliest = item

print("Total Bill :", total)
print("Most Expensive :", costliest)
print("Price :", high)
```
