```python
from functools import reduce

data = [
    [2, 3, 4],
    [5, 6, 7],
    [8, 9, 10]
]

result = reduce(
    lambda a, b: a + b,

    map(
        lambda row: reduce(
            lambda total, x:
                total + x**2 if x % 2 == 0 else total,
            row,
            0
        ),
        data
    )
)

print(result)
```
