```python
from functools import reduce

students = [
    {"name": "Manik", "marks": 80},
    {"name": "Taufiq", "marks": 90},
    {"name": "Abhi", "marks": 75}
]

total = reduce(
    lambda total, student: total + student["marks"],
    students,
    0
)

print(total)
print("Average:", total / len(students))
```
