```python
employees = (
    (101, ("Manik", "Developer", 50000)),
    (102, ("Rahul", "Tester", 42000)),
    (103, ("Sneha", "Manager", 75000))
)

for emp in employees:
    emp_id = emp[0]
    details = emp[1]

    print("Employee ID:", emp_id)
    print("Name:", details[0])
    print("Designation:", details[1])
    print("Salary:", details[2])
    print()
```
