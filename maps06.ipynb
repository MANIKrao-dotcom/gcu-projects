```python
# Python map() function - Long Example

# List of student names
names = ["Manik", "Rahul", "Arjun", "Kiran", "Suresh"]

# Marks of students in 5 subjects
maths = [85, 72, 90, 65, 78]
physics = [80, 75, 88, 70, 82]
chemistry = [82, 68, 92, 72, 76]
python = [90, 80, 95, 75, 85]
english = [78, 74, 89, 68, 80]


# Function to calculate total marks
def calculate_total(m, p, c, py, e):
    return m + p + c + py + e


# Using map() to calculate total marks
totals = list(map(calculate_total, maths, physics, chemistry, python, english))

print("Student Marks Report")
print("---------------------")

for i in range(len(names)):
    print(names[i], "Total =", totals[i])


# Function to calculate percentage
def calculate_percentage(total):
    return total / 5


# Using map() to calculate percentages
percentages = list(map(calculate_percentage, totals))

print("\nPercentage")
print("----------")

for i in range(len(names)):
    print(names[i], "=", percentages[i], "%")


# Function to find grade
def find_grade(percentage):

    if percentage >= 90:
        return "A+"

    elif percentage >= 80:
        return "A"

    elif percentage >= 70:
        return "B"

    elif percentage >= 60:
        return "C"

    elif percentage >= 50:
        return "D"

    else:
        return "F"


# Using map() to find grades
grades = list(map(find_grade, percentages))

print("\nGrades")
print("------")

for i in range(len(names)):
    print(names[i], "=", grades[i])


# Function to check pass/fail
def check_result(percentage):

    if percentage >= 40:
        return "PASS"

    else:
        return "FAIL"


# Using map() to check result
results = list(map(check_result, percentages))

print("\nResult")
print("------")

for i in range(len(names)):
    print(names[i], "=", results[i])


# Add bonus marks
def add_bonus(total):
    return total + 10


bonus_totals = list(map(add_bonus, totals))

print("\nTotal Marks After Bonus")
print("-----------------------")

for i in range(len(names)):
    print(names[i], "=", bonus_totals[i])


# Convert names to uppercase using map()
def convert_upper(name):
    return name.upper()


upper_names = list(map(convert_upper, names))

print("\nNames in Uppercase")
print("------------------")

for name in upper_names:
    print(name)


# Find length of each name
def name_length(name):
    return len(name)


lengths = list(map(name_length, names))

print("\nName Length")
print("-----------")

for i in range(len(names)):
    print(names[i], "=", lengths[i])


# Final report
print("\n==============================")
print("        FINAL REPORT")
print("==============================")

for i in range(len(names)):

    print("\nStudent Name :", names[i])
    print("Maths        :", maths[i])
    print("Physics      :", physics[i])
    print("Chemistry    :", chemistry[i])
    print("Python       :", python[i])
    print("English      :", english[i])
    print("Total        :", totals[i])
    print("Percentage   :", percentages[i], "%")
    print("Grade        :", grades[i])
    print("Result       :", results[i])
    print("Bonus Total  :", bonus_totals[i])
```
