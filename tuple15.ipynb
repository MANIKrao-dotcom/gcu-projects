```python
players = (
    ("Virat", (50, 70, 100)),
    ("Rohit", (80, 60, 40)),
    ("Gill", (90, 95, 88))
)

for player in players:
    name = player[0]
    scores = player[1]

    highest = max(scores)
    lowest = min(scores)
    average = sum(scores) / len(scores)

    print("Player:", name)
    print("Highest:", highest)
    print("Lowest:", lowest)
    print("Average:", average)
    print()
```
