SLICING IN LIST


```python
'''list[start:end]
the end index is  not included in the output'''
list =[10, 367, 890, 45, 0,"manik",  "bhadrakaalii"]
print(list[0:4])
print(list[1:5])
print(list[0:len(list)])
print(list[:len(list)])
```

FUNCTION OF LIST

```python
'''APPEND IS THE VARIABLE ADDED AT END OF LIST '''
list = [10, 40, 50, 64, 32, 2, "manik","bhairav"]
list.append("kaali")#this make changes in the list but if you run it wont give you output
                    #if you want the output then then print'''
print(list)
list.append("shiv")
print(list)
```

```python
'''REMOVE HERE REMOVE THE VALUE INN THE LIST
BUT WHEN YOU COME TO POP IT REMOVE THE ELEMENT BASED ON THE INDEX '''
list = [10, 40, 50, 64, 32, 2, "manik","bhairav"]
list.remove("manik")
print(list)
```

```python
'''INSERT IS THE FUNCTION WHICH INSERT THE VARIABLE IN A SPECIFIC INDEX
LIST[INDEX,VALUE]'''
list = [10, 40, 50, 64, 32, 2, "manik","bhairav"]
list.insert(1,"radha")
print(list)

```

```python
'''POP IS THE FUNCTION TO REMOVE THE VALUE FROM A SPECIFIC INDEX
IF YOU DINT MENTIONED THE SPECIFIC INDEX THEN IT WILL POP THE VALUE OF LAST INDEX'''
list = [10, 40, 50, 64, 32, 2, "manik","bhairav"]
list.pop()
print(list)
list.pop(2)
print(list)
```

```python
'''SORT IS A FUNCTION WHERE THE ELEMENTS ARE REARRANGED IN ORDER'''
list = [10, 40, 50, 64, 32, 2, "manik","bhairav"]

list.sort()#sort() did not work because your list contains both numbers and strings.
print(list)#because numbers and words are different data types.
```

```python
'''IN REVERSE() FUNCTION THE GIVEN VARIABLE IN LIST IS REVERSED
HERE THERE IS NO NEED TO DIFFERENCATE DATA TYPE AS SORT METHORD'''
list = [10, 40, 50, 64, 32, 2,]
list.reverse()
print(list)
```

```python
list = [10, 40, 50, 64, 32, 2,]
list.sort()
list.reverse()
print(list)
```
