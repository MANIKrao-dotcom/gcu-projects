```python
students = (
    ("Manik", (85, 90, 88)),
    ("Rahul", (75, 80, 78)),
    ("Anjali", (92, 95, 91))
)

for student in students:
    name = student[0]
    marks = student[1]

    total = sum(marks)
    average = total / len(marks)

    print("Name:", name)
    print("Marks:", marks)
    print("Total:", total)
    print("Average:", average)
    print("-" * 30)
```
