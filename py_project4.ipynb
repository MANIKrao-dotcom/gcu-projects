```python
import random

class Game:
    def __init__(self):
        self.number = random.randint(1, 100)
        self.attempts = 0

    def guess(self, num):
        self.attempts += 1

        if num < self.number:
            print("Too low!")
        elif num > self.number:
            print("Too high!")
        else:
            print("Correct!")
            print("Attempts:", self.attempts)


def start_game():
    game = Game()

    print("Guess a number between 1 and 100")

    while True:
        num = int(input("Enter your guess: "))
        game.guess(num)

        if num == game.number:
            break


start_game()
```
