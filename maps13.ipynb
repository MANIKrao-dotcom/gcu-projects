```python
students = ["Manik", "Taufiq", "Abhilash", "Suresh", "Rahul"]

maths = [85, 72, 91, 65, 78]
physics = [80, 75, 88, 70, 82]
python = [92, 68, 95, 72, 85]

total = list(map(
    lambda m, p, c: m + p + c,
    maths, physics, python
))

average = list(map(
    lambda x: x / 3,
    total
))

result = list(map(
    lambda x: "Distinction" if x >= 85
    else "First Class" if x >= 70
    else "Pass" if x >= 40
    else "Fail",
    average
))

print("Student Results")
print("--------------------------")

for i in range(len(students)):
    print("Name:", students[i])
    print("Maths:", maths[i])
    print("Physics:", physics[i])
    print("Python:", python[i])
    print("Total:", total[i])
    print("Average:", round(average[i], 2))
    print("Result:", result[i])
    print("--------------------------")
```
