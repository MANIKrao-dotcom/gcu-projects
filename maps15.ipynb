```python
products = [
    "Laptop",
    "Keyboard",
    "Mouse",
    "Monitor",
    "Headphones"
]

price = [55000, 1500, 800, 12000, 2500]
quantity = [1, 2, 3, 1, 2]

amount = list(map(
    lambda p, q: p * q,
    price, quantity
))

discount = list(map(
    lambda x: x * 0.10 if x >= 10000
    else x * 0.05 if x >= 5000
    else 0,
    amount
))

gst = list(map(
    lambda x: x * 0.18,
    amount
))

final_amount = list(map(
    lambda a, d, g: a - d + g,
    amount, discount, gst
))

print("SHOPPING BILL")
print("==============================")

for i in range(len(products)):
    print("Product:", products[i])
    print("Price:", price[i])
    print("Quantity:", quantity[i])
    print("Amount:", amount[i])
    print("Discount:", discount[i])
    print("GST:", gst[i])
    print("Final Amount:", final_amount[i])
    print("------------------------------")

print("Grand Total:", sum(final_amount))
```
